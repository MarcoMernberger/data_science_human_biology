# Setup Test - Human Biology Python Course

Dieses Notebook überprüft, ob die Python-Umgebung korrekt installiert wurde.

Es werden getestet:

- Python-Version
- wichtige Python-Pakete
- grundlegende Datenanalyse-Funktionen
- Visualisierung

Wenn am Ende die Erfolgsmeldung erscheint und die Grafik angezeigt wird,
ist die Installation erfolgreich.

```python
import sys
import platform

print("Python Setup Test")
print("=" * 50)

print(f"Python Version: {sys.version}")
print(f"Operating System: {platform.system()}")
print(f"Architecture: {platform.machine()}")

print("=" * 50)
```

```python
# Test der installierten Pakete

packages = {}

import numpy
packages["numpy"] = numpy.__version__

import pandas
packages["pandas"] = pandas.__version__

import matplotlib
packages["matplotlib"] = matplotlib.__version__

import scipy
packages["scipy"] = scipy.__version__

import sklearn
packages["scikit-learn"] = sklearn.__version__

import seaborn
packages["seaborn"] = seaborn.__version__

import statsmodels
packages["statsmodels"] = statsmodels.__version__

import Bio
packages["biopython"] = Bio.__version__

import pyarrow
packages["pyarrow"] = pyarrow.__version__

import openpyxl
packages["openpyxl"] = openpyxl.__version__

import requests
packages["requests"] = requests.__version__

import tqdm
packages["tqdm"] = tqdm.__version__

import IPython
packages["ipykernel/Jupyter"] = IPython.__version__


print("Installierte Pakete:")
print("=" * 50)

for package, version in packages.items():
    print(f"{package:20s} {version}")

print("=" * 50)
```

```python
# Test grundlegender numerischer Berechnungen

import numpy as np

data = np.array([1, 2, 3, 4, 5])

print("Testdaten:")
print(data)

print()

print("Mittelwert:")
print(np.mean(data))

print()

print("Standardabweichung:")
print(np.std(data))
```

```python
# Test mit pandas

import pandas as pd

df = pd.DataFrame(
    {
        "sample": ["A", "B", "C"],
        "measurement": [12.3, 15.7, 11.9],
    }
)

df
```

```python
# Test Visualisierung

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()

plt.figure(figsize=(6, 4))

sns.barplot(
    data=df,
    x="sample",
    y="measurement",
)

plt.title("Test Plot - Human Biology Course")
plt.ylabel("Measurement")
plt.xlabel("Sample")

plt.show()
```
