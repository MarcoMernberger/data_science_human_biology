### 4. How does Python execute code?

Computers understand machine instructions.

Humans write readable code.

The Python interpreter translates our instructions into something the computer can execute.

```python
Human readable code

        ↓

Python interpreter

        ↓

Computer instructions

        ↓

Output
```

Python is called an interpreted language because code can be executed step by step.

If we want to do something with values, we need to know the <font color='green'>**operators**</font>
